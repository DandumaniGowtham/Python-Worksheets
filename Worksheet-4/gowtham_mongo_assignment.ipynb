# MongoDB Assignment
## Part - 0  Installation and Connection to MongoDB

```python
from pymongo import MongoClient   # Import MongoClient from PyMongo to connect Python with MongoDB

client = MongoClient("mongodb://localhost:27017/")  # Connect to the local MongoDB server
db = client["school_assignment"] # Access the school_assignment database

students = db["students"] # Giving access to students collection
courses = db["courses"]  # Access the courses collection
enrollments = db["enrollments"]  # Access the enrollments collection
```

#  Part - 1

```python
# 1.1 -  Show Evey Document in Student Collection
for doc in students.find({}):
    print(doc)
```

```python
# 1.2  Show only name and city of students who are older than 10
result = students.find(
    {"age": {"$gt": 10}},
    {"_id": 0, "name": 1, "city": 1}
)

for doc in result:
    print(doc)
```

```python
# 1.3 — Students in grade 5 or grade 7, sorted by name A → Z
result = students.find(
    {"$or": [{"grade": 5}, {"grade": 7}]}
).sort("name", 1)

for doc in result:
    print(doc)
```

```python
# 1.4 — Students whose name starts with A
result = students.find(
    {"name": {"$regex": "^A"}}
)

for doc in result:
    print(doc)
```

```python
# 1.5 — Find students whose marks are missing
result = students.find(
    {"marks": None}
)

for doc in result:
    print(doc)
```

```python
# 1.6 — Top 3 students by marks, highest first
result = students.find(
    {"marks": {"$ne": None}}
).sort("marks", -1).limit(3)

for doc in result:
    print(doc)
```

```python
# 1.7 — Show distinct cities
result = students.distinct("city")

for city in result:
    print(city)
```

```python
# 1.8 — Students who know both math AND science
result = students.find(
    {"skills": {"$all": ["math","science"]}}
)

for doc in result:
    print(doc)
```

```python
# 1.9 — Students whose skills array has exactly 2 elements
result = students.find(
    {"skills": {"$size": 2}}
)

for doc in result:
    print(doc)
```

```python
# 1.10 — INSERT → UPDATE → DELETE

# INSERT Priya
result = students.insert_one({
    "name": "Priya",
    "age": 11,
    "grade": 6,
    "city": "Pune",
    "marks": 81,
    "skills": ["art"]
})

# UPDATE Priya's marks to 85
students.update_one(
    {"name": "Priya"},
    {"$set": {"marks": 85}}
)

# DELETE Priya
students.delete_one(
    {"name": "Priya"}
)
```

# Part 2 — Update & Delete 

```python
# 2.1 — Add email to Kabir using $set
students.update_one(
    {"name": "Kabir"},
    {"$set": {"email": "kabir@gmail.com"}}
)
```

```python
# 2.2 — Rename email to email_id
students.update_one(
    {"name": "Kabir"},
    {"$rename": {"email": "email_id"}}
)
```

```python
# 2.3 Use $unset to remove email_id from Kabir
students.update_one(
    {"name": "Kabir"},
    {"$unset": {"email_id": ""}}
)
```

```python
# 2.4 Use $inc to give +3 marks to every grade 5 student.
students.update_many(
    {"grade": 5},
    {"$inc": {"marks": 3}}
)
```

```python
# 2.5 — Add "coding" to all Mumbai students without duplicates
students.update_many(
    {"city": "Mumbai"},
    {"$addToSet": {"skills": "coding"}}
)
```

```python
# 2.6  Use upsert=True to update student "Ishaan" — set city Delhi, grade 7, marks 60.
result = students.update_one(
    {"name": "Ishaan"},
    {"$set": {"city": "Delhi", "grade": 7, "marks": 60}},
    upsert=True # if document found it will update, else it creates a new document
)

print(result.upserted_id)  # If new Document is created, this prints ID of the Document
```

```python
# 2.7 — Delete Ishaan, then delete students with marks below 58
students.delete_one(
    {"name": "Ishaan"}
)

students.delete_many(
    {"marks": {"$lt": 58}}
)
```

# Part 3 — Aggregation Basics. 

```python
# 3.1 — Number of students and average age
result = students.aggregate([
    {
        "$group": {
            "_id": None,
            "total": {"$sum": 1},
            "avgAge": {"$avg": "$age"}
        }
    }
])

for doc in result:
    print(doc)
```

```python
# 3.2 — Highest and lowest marks
result = students.aggregate([
    {
        "$group": {
            "_id": None,
            "max": {"$max": "$marks"},
            "min": {"$min": "$marks"}
        }
    }
])

for doc in result:
    print(doc)
```

```python
# 3.3 — Compare total documents with documents having non-null marks
print(students.count_documents({}))

print(students.count_documents({
    "marks": {"$ne": None}
}))
```

```python
# 3.4 — Count students in each grade, sorted by grade
result = db.students.aggregate([
    {"$group": {"_id": "$grade", "count": {"$sum": 1}}},
    {"$match": {"count": {"$gte": 2}}},
    {"$sort": {"_id": 1}}
])

for doc in result:
    print(doc)
```

```python
# 3.5 — Average marks per city, rounded to 1 decimal, highest city first
result = db.students.aggregate([
    {"$match": {"marks": {"$ne": None}}},
    {"$group": {"_id": "$city", "avg": {"$avg": "$marks"}}},
    {"$project": {"avg": {"$round": ["$avg", 1]}}},
    {"$sort": {"avg": -1}}
])

for doc in result:
    print(doc)
```

```python
# 3.6 — Only cities whose averages are above 75
result = students.aggregate([
    {"$match": {"marks": {"$ne": None}}},
    {"$group": {"_id": "$city", "avg": {"$avg": "$marks"}}},
    {"$project": {"avg": {"$round": ["$avg", 1]}}},
    {"$match": {"avg": {"$gt": 75}}},
    {"$sort": {"avg": -1}}
])

for res in result:
    print(res)
```

```python
# 3.7 — Grades that have at least 2 students
result = db.students.aggregate([
    {"$group": {"_id": "$grade", "count": {"$sum": 1}}},
    {"$match": {"count": {"$gte": 2}}}
])

for doc in result:
    print(doc)
```

```python
# 3.8 — Count students in each course_id
result = enrollments.aggregate([
    {"$group": {"_id": "$course_id", "student_count": {"$sum": 1}}},
    {"$sort": {"_id": 1}}
])

for doc in result:
    print(doc)
```

```python
# 3.9 — Highest score in each course_id
result = enrollments.aggregate([
    {"$group": {"_id": "$course_id", "highest_score": {"$max": "$score"}}},
    {"$sort": {"_id": 1}}
])

for doc in result:
    print(doc)
```

```python
# 3.10 — Number of students AND number who have marks, for each city
result = students.aggregate([
    {"$group": {"_id": "$city", "total": {"$sum": 1},
                "withMarks": {"$sum": {"$cond": [
                    {"$ne": ["$marks", None]}, 1, 0
                ]}}}}
])

for doc in result:
    print(doc)
```

# Part 4 — Aggregation Advanced

```python
# 4.1 — Join enrollments and  courses, List student_id, course_name, score.
result = enrollments.aggregate([
    {"$lookup": {
        "from": "courses",
        "localField": "course_id",
        "foreignField": "_id",
        "as": "c"
    }},
    {"$unwind": "$c"},
    {"$project": {
        "_id": 0,
        "student_id": 1,
        "course_name": "$c.course_name",
        "score": 1
    }}
])

for res in result:
    print(res)
```

```python
# 4.2 — Join all three collections# 
result = db.students.aggregate([
    {"$lookup": {
        "from": "enrollments",
        "localField": "_id",
        "foreignField": "student_id",
        "as": "e"
    }},
    {"$unwind": {
        "path": "$e",
        "preserveNullAndEmptyArrays": True
    }},
    {"$lookup": {
        "from": "courses",
        "localField": "e.course_id",
        "foreignField": "_id",
        "as": "c"
    }},
    {"$unwind": {
        "path": "$c",
        "preserveNullAndEmptyArrays": True
    }},
    {"$project": {
        "_id": 0,
        "name": 1,
        "course_name": "$c.course_name",
        "score": "$e.score"
    }}
])

for res in result:
 print (res)
```

```python
# 4.3 — Students not enrolled in any course
result = students.aggregate([
    {"$lookup": {
        "from": "enrollments",
        "localField": "_id",
        "foreignField": "student_id",
        "as": "e"
    }},
    {"$match": {
        "e": {"$size": 0}
    }}
])

for res in result:
    print(res)
```

```python
# 4.4 — Every course with number of students, including courses with 0
result = courses.aggregate([
    {"$lookup": {
        "from": "enrollments",
        "localField": "_id",
        "foreignField": "course_id",
        "as": "e"
    }},
    {"$project": {
        "course_name": 1,
        "studentCount": {"$size": "$e"}
    }}
])

for res in result:
    print(res)
```

```python
# 4.5 — Each student and how many courses they take
result = db.students.aggregate([
    {"$lookup": {
        "from": "enrollments",
        "localField": "_id",
        "foreignField": "student_id",
        "as": "e"
    }},
    {"$project": {
        "name": 1,
        "courseCount": {"$size": "$e"}
    }},
    {"$sort": {
        "courseCount": -1,
        "name": 1
    }}
])

for res in result:
    print(res)
```

```python
# 4.7 — Label students based on marks
result = students.aggregate([
    {"$addFields": {
        "label": {
            "$switch": {
                "branches": [
                    {"case": {"$eq": ["$marks", None]}, "then": "No marks"},
                    {"case": {"$gte": ["$marks", 85]}, "then": "Excellent"},
                    {"case": {"$gte": ["$marks", 70]}, "then": "Good"}
                ],
                "default": "Needs practice"
            }
        }
    }}
])

for res in result:
    print(res)
```

```python
# 4.8 — Average score per course name
result = enrollments.aggregate([
    {"$match": {"score": {"$ne": None}}},
    {"$lookup": {"from": "courses", "localField": "course_id", "foreignField": "_id", "as": "c"}},
    {"$unwind": "$c"},
    {"$group": {"_id": "$c.course_name", "avg": {"$avg": "$score"}}},
    {"$project": {"avg": {"$round": ["$avg", 1]}}},
    {"$sort": {"avg": -1}}
])

for res in result:
    print(res)
```

```python
# 4.9 — Self-lookup: students living in the same city
result = students.aggregate([
    {"$lookup": {
        "from": "students",
        "let": {"city": "$city", "id": "$_id"},
        "pipeline": [
            {"$match": {
                "$expr": {
                    "$and": [
                        {"$eq": ["$city", "$$city"]},
                        {"$lt": ["$$id", "$_id"]}
                    ]
                }
            }}
        ],
        "as": "partner"
    }},
    {"$unwind": "$partner"}
])

for doc in result:
    print(doc)
```

```python
 # 4.10 — $facet: 3 results in one pipeline
result = students.aggregate([
    {"$facet": {
        "byGrade": [
            {"$group": {"_id": "$grade", "count": {"$sum": 1}}}
        ],
        "byCity": [
            {"$group": {"_id": "$city", "count": {"$sum": 1}}}
        ],
        "topScorers": [
            {"$match": {"marks": {"$ne": None}}},
            {"$sort": {"marks": -1}},
            {"$limit": 2}
        ]
    }}
])

for res in result:
    print(res)
```

# Part 6 — Bonus Challenges

```python
# 1 — Teacher with the most students across all courses
result = courses.aggregate([
    {"$lookup": {
        "from": "enrollments",
        "localField": "_id",
        "foreignField": "course_id",
        "as": "e"
    }},
    {"$project": {
        "teacher": 1,
        "count": {"$size": "$e"}
    }},
    {"$group": {
        "_id": "$teacher",
        "total": {"$sum": "$count"}
    }}
])

for res in result:
    print(res)
```

```python
# 2 - Difference between best enrollment score and student's marks
result = students.aggregate([
    {"$lookup": {"from": "enrollments", "localField": "_id", "foreignField": "student_id", "as": "e"}},
    {"$match": {"e": {"$ne": []}, "marks": {"$ne": None}}},
    {"$project": {"name": 1, "marks": 1, "bestScore": {"$max": "$e.score"}}},
    {"$addFields": {"improvement": {"$subtract": ["$bestScore", "$marks"]}}}
])

for doc in result:
    print(doc)
```

```python
# 6.3 — Save average marks per city into city_stats
result = students.aggregate([
    {"$group": {"_id": "$city", "avg_marks": {"$avg": "$marks"}}},
    {"$project": {"_id": 0, "city": "$_id", "avg_marks": {"$round": ["$avg_marks", 1]}}},
    {"$out": "city_stats"}
])

for doc in db.city_stats.find({}):
    print(doc)
```
